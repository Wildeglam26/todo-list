```python
# Sistema de Gestão de Notas de Alunos
# Disciplina: Linguagem de Programação
# Unidade: U1 - Introdução à Linguagem Python
# Aula: A4 - Funções em Python

# Lista que armazenará as notas
notas = []

print("===== SISTEMA DE GESTÃO DE NOTAS DE ALUNOS =====")
print("Digite as notas do aluno.")
print("Digite -1 para finalizar o cadastro.\n")

# Cadastro das notas
while True:
    nota = float(input("Digite uma nota: "))

    # Verifica se o usuário deseja finalizar
    if nota == -1:
        break

    # Adiciona a nota à lista
    notas.append(nota)


# Função responsável pelo cálculo da média
def calcular_media(notas):
    return sum(notas) / len(notas)


# Função responsável por determinar a situação do aluno
def determinar_situacao(media):
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"


# Função responsável por exibir o relatório final
def exibir_relatorio(notas, media, situacao):
    print("\n===== RELATÓRIO FINAL =====")
    print("Notas cadastradas:", notas)
    print("Média:", round(media, 2))
    print("Situação:", situacao)


# Verifica se existem notas cadastradas
if len(notas) > 0:

    # Calcula a média
    media = calcular_media(notas)

    # Determina a situação
    situacao = determinar_situacao(media)

    # Exibe o relatório
    exibir_relatorio(notas, media, situacao)

else:
    print("Nenhuma nota foi cadastrada.")
```
